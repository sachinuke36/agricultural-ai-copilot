```python
train_df = df_yield[df_yield["Crop_Year"] <= 2017].copy()

test_df = df_yield[df_yield["Crop_Year"] >= 2018].copy()

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)

print("\nTraining years:")
print(train_df["Crop_Year"].min(), "-", train_df["Crop_Year"].max())

print("\nTesting years:")
print(test_df["Crop_Year"].min(), "-", test_df["Crop_Year"].max())
```

```python
features = [
    "Crop",
    "Crop_Year",
    "Season",
    "State",
    "Area",
    "Annual_Rainfall",
    "Fertilizer",
    "Pesticide",
    "Avg_Temperature",
    "Max_Temperature",
    "Min_Temperature"
]

target = "Yield"

X_train = train_df[features]
y_train = train_df[target]

X_test = test_df[features]
y_test = test_df[target]
```

```python
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
unseen_crops = set(X_test["Crop"]) - set(X_train["Crop"])
unseen_states = set(X_test["State"]) - set(X_train["State"])

print("Unseen crops:", unseen_crops)
print("Unseen states:", unseen_states)
```

```python
print(
    df_yield.nlargest(10, "Yield")[
        ["Crop", "Crop_Year", "State", "Area", "Production", "Yield"]
    ]
)
```

```python
categorical_features = [
    "Crop",
    "Season",
    "State"
]

numerical_features = [
    "Crop_Year",
    "Area",
    "Annual_Rainfall",
    "Fertilizer",
    "Pesticide",
    "Avg_Temperature",
    "Max_Temperature",
    "Min_Temperature"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)
```

```python
import numpy as np

y_train_log = np.log1p(y_train)
y_test_log = np.log1p(y_test)
```

```python
from sklearn.ensemble import RandomForestRegressor

regressor = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)
```

```python
from sklearn.pipeline import Pipeline

pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", regressor)
    ]
)
```

```python
pipeline.fit(
    X_train,
    y_train_log
)
```

```python
y_pred_log = pipeline.predict(X_test)
```

```python
y_pred = np.expm1(y_pred_log)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

r2 = r2_score(
    y_test,
    y_pred
)

print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
results = test_df[
    ["Crop", "Crop_Year", "State", "Yield"]
].copy()

results["Predicted_Yield"] = y_pred

results["Absolute_Error"] = (
    results["Yield"] - results["Predicted_Yield"]
).abs()

results.head(10)
```

```python
worst_predictions = results.sort_values(
    "Absolute_Error",
    ascending=False
)

worst_predictions.head(15)
```

```python
crop_performance = (
    results
    .groupby("Crop")
    .agg(
        Actual_Mean=("Yield", "mean"),
        Predicted_Mean=("Predicted_Yield", "mean"),
        MAE=("Absolute_Error", "mean"),
        Samples=("Yield", "count")
    )
    .sort_values("MAE", ascending=False)
)

crop_performance
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))

plt.scatter(
    results["Yield"],
    results["Predicted_Yield"],
    alpha=0.4
)

plt.xlabel("Actual Yield")
plt.ylabel("Predicted Yield")
plt.title("Actual vs Predicted Crop Yield")

plt.show()
```

```python
baseline_prediction = np.full(
    len(y_test),
    y_train.mean()
)

baseline_mae = mean_absolute_error(
    y_test,
    baseline_prediction
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        baseline_prediction
    )
)

baseline_r2 = r2_score(
    y_test,
    baseline_prediction
)

print("Baseline MAE :", baseline_mae)
print("Baseline RMSE:", baseline_rmse)
print("Baseline R²  :", baseline_r2)
```

```python
from sklearn.metrics import mean_absolute_percentage_error

mape = mean_absolute_percentage_error(
    y_test,
    y_pred
)

print("MAPE:", mape)
print("MAPE (%):", mape * 100)
```

```python
mask = y_test > 0

mape_nonzero = mean_absolute_percentage_error(
    y_test[mask],
    y_pred[mask]
)

print("MAPE excluding zero yields:", mape_nonzero)
print("MAPE (%):", mape_nonzero * 100)
```

```python
results["Percentage_Error"] = (
    results["Absolute_Error"] /
    results["Yield"].replace(0, np.nan)
) * 100
```

```python
results[
    ["Crop", "Crop_Year", "State",
     "Yield", "Predicted_Yield",
     "Absolute_Error", "Percentage_Error"]
].head(20)
```

```python
crop_percentage_error = (
    results
    .groupby("Crop")
    .agg(
        MAPE=("Percentage_Error", "mean"),
        MAE=("Absolute_Error", "mean"),
        Samples=("Yield", "count")
    )
    .sort_values("MAPE", ascending=False)
)

crop_percentage_error.head(20)
```

```python
feature_names = pipeline.named_steps[
    "preprocessor"
].get_feature_names_out()

print("Number of features:", len(feature_names))
```

```python
import pandas as pd

importances = pipeline.named_steps[
    "regressor"
].feature_importances_

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

feature_importance.head(20)
```

```python
import matplotlib.pyplot as plt

top_features = feature_importance.head(15)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top Features Influencing Yield Prediction")

plt.show()
```

```python
raw_regressor = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

raw_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", raw_regressor)
    ]
)

raw_pipeline.fit(
    X_train,
    y_train
)
```

```python
y_pred_raw = raw_pipeline.predict(X_test)
```

```python
raw_mae = mean_absolute_error(
    y_test,
    y_pred_raw
)

raw_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_raw
    )
)

raw_r2 = r2_score(
    y_test,
    y_pred_raw
)

print("Raw Target Model")
print("MAE :", raw_mae)
print("RMSE:", raw_rmse)
print("R²  :", raw_r2)
```
